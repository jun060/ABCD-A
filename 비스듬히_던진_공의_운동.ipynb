```python
import math

# 초기 값
v0 = 100  # 초기 속도 (m/s)
theta_deg = 37  # 발사 각도 (degrees)
y0 = 140  # 초기 높이 (m)
g = 9.8  # 중력 가속도 (m/s^2)

# 각도를 라디안으로 변환
theta_rad = math.radians(theta_deg)

# 속도 성분
v0x = v0 * math.cos(theta_rad)
v0y = v0 * math.sin(theta_rad)

# (1) 시간 구하기 (y = y0 + v0y * t - 0.5 * g * t^2 = 0)
# 0 = y0 + v0y * t - 0.5 * g * t^2  → 이차방정식
a = -0.5 * g
b = v0y
c = y0

discriminant = b**2 - 4*a*c
if discriminant < 0:
    print("해가 없다.")
else:
    t1 = (-b + math.sqrt(discriminant)) / (2*a)
    t2 = (-b - math.sqrt(discriminant)) / (2*a)
    t = max(t1, t2)  # 양의 시간 선택
    print(f"(a) 충돌까지 걸린 시간: {t:.2f}초")

    # (2) 수평 거리
    x = v0x * t
    print(f"(b) 수평 거리: {x:.2f}m")

    # (3) 충돌 시 순간 속도 성분
    vx = v0x  # 수평 속도는 일정
    vy = v0y - g * t
    print(f"(c) 충돌 시 순간 속도 성분: vx = {vx:.2f} m/s, vy = {vy:.2f} m/s")

    # (4) 속력
    v = math.sqrt(vx**2 + vy**2)
    print(f"(d) 충돌 시 속력: {v:.2f} m/s")

    # (5) 충돌 시 속도 방향 각도 (아래 방향으로 기울어짐)
    angle_rad = math.atan2(vy, vx)
    angle_deg = math.degrees(angle_rad)
    print(f"(e) 충돌 시 속도 방향 각도: {angle_deg:.2f}도")
```
